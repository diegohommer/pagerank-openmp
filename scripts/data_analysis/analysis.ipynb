```python
import os
import csv
import pandas as pd
from pathlib import Path

def load_results_dataframe(base_dir: Path) -> dict:
    """
    Reads result.csv files from subdirectories of base_dir
    and builds a pandas DataFrame:
    
         Elapsed Time     CPI Rate  ...  Operating System
    run1     10.650258     0.551805  ...  6.1.0-28-amd64
    run2     ...          ...       ...  ...
    """
    data = {}

    for subdir in os.listdir(base_dir):
        subdir_path = os.path.join(base_dir, subdir)
        result_file = os.path.join(subdir_path, "report.csv")

        if not os.path.isdir(subdir_path):
            continue
        if not os.path.exists(result_file):
            continue

        fields = {}
        with open(result_file, newline='', encoding="utf-8") as f:
            reader = csv.reader(f, delimiter="\t")
            for row in reader:
                if len(row) < 2:
                    continue
                key = row[-2].strip()
                val = row[-1].strip()
                if key and val:
                    fields[key] = val

        data[subdir] = fields

    return data
```

```python
print(load_results_dataframe(Path("../../data/results/")))
```
